### 2.Quartile and Interquartile Range (IQR) Analysis

Quartiles divide the data into four parts and help describe the distribution of numerical variables.

Q1 (25th percentile): Approximately 25% of the observations are below this value.
Q2 (50th percentile): This is the median, where approximately 50% of the observations are below this value.
Q3 (75th percentile): Approximately 75% of the observations are below this value.

The Interquartile Range (IQR) measures the spread of the middle 50% of the data.

The formula is:

IQR = Q3 − Q1

The IQR is useful for understanding variability and identifying potential outliers.

In this analysis, Q1, Q3, IQR, and potential outlier bounds are calculated for the numerical variables.

```python
# QUARTILE AND IQR ANALYSIS

print("\n" + "=" * 70)
print("QUARTILE AND INTERQUARTILE RANGE (IQR) ANALYSIS")
print("=" * 70)

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q2 = df[column].quantile(0.50)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ][column].count()

    print(f"\n{column}")
    print("Q1 (25%):", round(Q1, 2))
    print("Q2 (Median):", round(Q2, 2))
    print("Q3 (75%):", round(Q3, 2))
    print("IQR:", round(IQR, 2))
    print("Lower Bound:", round(lower_bound, 2))
    print("Upper Bound:", round(upper_bound, 2))
    print("Potential Outliers:", outliers)
```

Interpretation of Quartile and IQR Analysis

The quartile analysis describes the distribution and spread of the numerical variables in the retail dataset.

Q1 (25th percentile) represents the value below which approximately 25% of observations fall, while Q3 (75th percentile) represents the value below which approximately 75% of observations fall. The difference between Q3 and Q1 is the Interquartile Range (IQR), which represents the spread of the middle 50% of the observations.

The IQR method was also used to calculate potential outlier boundaries using:

Lower Bound = Q1 − 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these boundaries are identified as potential outliers according to the IQR rule.

For Product Price, Q1 is approximately ₹15,375, Q3 is approximately ₹44,957, and the IQR is approximately ₹29,582. The analysis identifies 0 potential outliers for Product Price using the IQR method.

For Final Price, Q1 is approximately ₹24,705 and Q3 is approximately ₹94,046.07, giving an IQR of approximately ₹69,341.07. The larger IQR indicates greater variability in Final Price compared with Product Price.

Overall, quartiles and IQR provide additional information about the spread of the data and help identify observations that may require further investigation.

### 3. Skewness and Kurtosis

Skewness and kurtosis are statistical measures used to understand the shape and distribution of numerical variables.

Skewness measures the degree and direction of asymmetry in a distribution.

A skewness value close to 0 indicates a relatively symmetric distribution.
A positive skewness value indicates that the distribution has a longer tail toward higher values.
A negative skewness value indicates that the distribution has a longer tail toward lower values.

Kurtosis describes the characteristics of the tails and the concentration of values around the center of a distribution. It can help identify whether a distribution has relatively heavier or lighter tails compared with a normal distribution.

Skewness and kurtosis are calculated for:

Age
Product Price
Quantity
Discount Percentage
Final Price
Delivery Days

These measures provide additional information about the distribution of the numerical variables and help determine whether further statistical analysis or visualization may be useful.

```python
# SKEWNESS AND KURTOSIS

distribution_stats = pd.DataFrame({
    "Skewness": df[numerical_columns].skew(),
    "Kurtosis": df[numerical_columns].kurt()
})

print("\n" + "=" * 70)
print("SKEWNESS AND KURTOSIS")
print("=" * 70)

print(distribution_stats.round(3))
```

Interpretation of Skewness and Kurtosis

The skewness results show that Age, Product Price, Quantity, Discount Percentage, and Delivery Days have values very close to zero. This indicates that their distributions are approximately symmetric.

The Final Price variable has a skewness value of 0.935, indicating a positive or right-skewed distribution. This means that the distribution has a longer tail toward higher final-price values. The difference between the mean Final Price (₹64,165.72) and median Final Price (₹49,752.39) is also consistent with this right-skewed distribution.

The kurtosis results provide additional information about the shape and tail behavior of the distributions. Age, Product Price, Quantity, Discount Percentage, and Delivery Days have negative excess kurtosis values, while Final Price has a value closer to zero.

These results help describe the distributional characteristics of the variables and provide useful context for interpreting the statistical analyses and visualizations that follow.

### 4. Confidence Interval Estimation

A confidence interval is used to estimate a range of plausible values for a population parameter based on a sample or observed dataset.

In this project, a 95% confidence interval is calculated for the mean Final Price.

The confidence interval provides a lower and upper bound around the estimated mean. A 95% confidence interval is commonly used in statistical analysis to quantify the uncertainty associated with estimating the population mean.

The calculation uses:

Mean Final Price
Standard deviation
Number of observations
Standard error
t-distribution

The resulting interval provides an estimate of the range within which the population mean Final Price may plausibly lie, subject to the assumptions of the statistical method.

```python
# 95% CONFIDENCE INTERVAL FOR MEAN FINAL PRICE

from scipy import stats

final_price = df["final_price"].dropna()

mean_final_price = final_price.mean()
std_final_price = final_price.std()
n = len(final_price)

standard_error = std_final_price / np.sqrt(n)

confidence_interval = stats.t.interval(
    confidence=0.95,
    df=n - 1,
    loc=mean_final_price,
    scale=standard_error
)

print("\n" + "=" * 70)
print("95% CONFIDENCE INTERVAL FOR MEAN FINAL PRICE")
print("=" * 70)

print("Mean Final Price:", round(mean_final_price, 2))
print("Lower Bound:", round(confidence_interval[0], 2))
print("Upper Bound:", round(confidence_interval[1], 2))
```

Interpretation of Confidence Interval

The estimated mean Final Price in the dataset is ₹64,165.72.

A 95% confidence interval for the mean Final Price was calculated as ₹63,856.01 to ₹64,475.43.

This interval provides a range of plausible values for the population mean Final Price based on the observed data and the assumptions of the statistical method.

The relatively narrow width of the interval is consistent with the large number of observations in the dataset (100,000 transactions), which provides a precise estimate of the mean.

The confidence interval complements the descriptive statistics by providing information about the uncertainty associated with estimating the mean Final Price.

### 5. Correlation Analysis

Correlation analysis is used to measure the strength and direction of the linear relationship between two numerical variables.

In this project, Pearson correlation is used to examine the relationship between Product Price and Final Price.

The Pearson correlation coefficient ranges from -1 to +1:

A value close to +1 indicates a strong positive linear relationship.
A value close to -1 indicates a strong negative linear relationship.
A value close to 0 indicates little or no linear relationship.

The correlation coefficient found earlier for Product Price and Final Price was approximately 0.731, indicating a strong positive linear association in this dataset.

A statistical significance test is also performed to determine whether the observed correlation is statistically distinguishable from zero at the 5% significance level.

```python
# PEARSON CORRELATION SIGNIFICANCE TEST

correlation, p_value = stats.pearsonr(
    df["product_price"],
    df["final_price"]
)

print("\n" + "=" * 70)
print("PEARSON CORRELATION TEST")
print("=" * 70)

print("Correlation coefficient:", round(correlation, 3))
print("p-value:", p_value)

if p_value < 0.05:
    print("The correlation is statistically significant at the 5% level.")
else:
    print("The correlation is not statistically significant at the 5% level.")
```

Interpretation of Correlation and Statistical Significance

The Pearson correlation coefficient between Product Price and Final Price is 0.731. This indicates a strong positive linear association in the analyzed dataset, meaning that higher Product Price values tend to be associated with higher Final Price values.

The Pearson correlation significance test produced a p-value displayed as 0.0. This indicates that the p-value is extremely small and is below the 5% significance level used in this analysis.

Therefore, the observed correlation is statistically significant at the 5% level. This provides statistical evidence that the observed linear association between Product Price and Final Price is distinguishable from zero in this dataset.

However, statistical significance does not by itself establish a causal relationship. The correlation describes an association between the two variables rather than proving that changes in Product Price directly cause changes in Final Price.
